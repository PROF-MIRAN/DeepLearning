# 0. 기본 코드 실행
아래 코드는 실험에 사용할 Iris 데이터를 준비합니다.


```python
import numpy as np
from sklearn import datasets
from sklearn.preprocessing import StandardScaler

# Iris Dataset
iris = datasets.load_iris()
X_all = iris.data
Y_all = iris.target

# Setosa(0), Versicolor(1)만 선택
mask = (Y_all == 0) | (Y_all == 1)
X_raw = X_all[mask].copy()
Y = Y_all[mask].copy()

# Label 변환: Setosa -> +1, Versicolor -> -1
Y = np.where(Y == 0, 1, -1)

print("X_raw shape:", X_raw.shape)
print("Y shape:", Y.shape)
```

## 공통 평가 함수

아래 함수는 Weight, Bias, 입력 데이터가 주어졌을 때  
Accuracy, MSE, MAE를 계산합니다.


```python
def evaluate_perceptron(X, Y, w, b):
    z = X @ w + b
    y_pred = np.where(z >= 0, 1, -1)

    accuracy = np.mean(y_pred == Y)
    mse = np.mean((Y - y_pred) ** 2)
    mae = np.mean(np.abs(Y - y_pred))

    return accuracy, mse, mae, y_pred
```

# 1. Weight 변화에 따른 성능 비교

기본 Weight와 변경 Weight를 비교하세요.

- 기본: `[0.5, 0.5, 0.5, 0.5]`
- 변경: `[0.1, 0.1, 0.1, 0.1]`


```python
# StandardScaler 적용
scaler = StandardScaler()
X = scaler.fit_transform(X_raw)

# TODO: 두 Weight 설정
w1 = np.array([0.5, 0.5, 0.5, 0.5])
w2 = np.array([0.1, 0.1, 0.1, 0.1])

b = 0.1

result1 = evaluate_perceptron(X, Y, w1, b)
result2 = evaluate_perceptron(X, Y, w2, b)

print("[Weight 1]")
print(f"Accuracy: {result1[0]:.3f}, MSE: {result1[1]:.3f}, MAE: {result1[2]:.3f}")

print("\n[Weight 2]")
print(f"Accuracy: {result2[0]:.3f}, MSE: {result2[1]:.3f}, MAE: {result2[2]:.3f}")
```

| `[0.5, 0.5, 0.5, 0.5]` |  |  |  |
| `[0.1, 0.1, 0.1, 0.1]` |  |  |  |

# 2. Bias 변화에 따른 성능 비교

다음 두 Bias를 비교하세요.

- 기본 Bias: `0.1`
- 변경 Bias: `-0.5`


```python
w = np.array([0.5, 0.5, 0.5, 0.5])

b1 = 0.1
b2 = -0.5

result1 = evaluate_perceptron(X, Y, w, b1)
result2 = evaluate_perceptron(X, Y, w, b2)

print("[Bias = 0.1]")
print(f"Accuracy: {result1[0]:.3f}, MSE: {result1[1]:.3f}, MAE: {result1[2]:.3f}")

print("\n[Bias = -0.5]")
print(f"Accuracy: {result2[0]:.3f}, MSE: {result2[1]:.3f}, MAE: {result2[2]:.3f}")
```

# 3. StandardScaler 적용 여부 비교

표준화를 적용한 경우와 적용하지 않은 경우를 비교하세요.


```python
w = np.array([0.5, 0.5, 0.5, 0.5])
b = 0.1

# StandardScaler 적용
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

# StandardScaler 미적용
X_unscaled = X_raw.copy()

scaled_result = evaluate_perceptron(X_scaled, Y, w, b)
unscaled_result = evaluate_perceptron(X_unscaled, Y, w, b)

print("[StandardScaler 적용]")
print(f"Accuracy: {scaled_result[0]:.3f}, MSE: {scaled_result[1]:.3f}, MAE: {scaled_result[2]:.3f}")
print("평균:", X_scaled.mean(axis=0))
print("표준편차:", X_scaled.std(axis=0))

print("\n[StandardScaler 미적용]")
print(f"Accuracy: {unscaled_result[0]:.3f}, MSE: {unscaled_result[1]:.3f}, MAE: {unscaled_result[2]:.3f}")
print("평균:", X_unscaled.mean(axis=0))
print("표준편차:", X_unscaled.std(axis=0))
```
