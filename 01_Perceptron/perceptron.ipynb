```python
# Code Cell [1]. 라이브러리

import numpy as np

from sklearn import datasets
from sklearn.preprocessing import StandardScaler
```

```python
# Code Cell [2]. iris 데이터 

iris = datasets.load_iris()

X_all = iris.data
Y_all = iris.target

print(iris.keys())
```

```python
# Code Cell [3]. 데이터 확인 

print("X shape:", X_all.shape)
print("Y shape:", Y_all.shape)

print("\nFeature Names")
print(iris.feature_names)

print("\nTarget Names")
print(iris.target_names)
```

```python
# Code Cell [4]. 실제 데이터 확인 

X_all[:10]
Y_all
```

```python
# Code Cell [5] Setosa / Versicolor 추출
   # Setosa(0)와 Versicolor(1)만 사용

mask = (Y_all == 0) | (Y_all == 1)

X = X_all[mask]
Y = Y_all[mask]

print(X.shape)
print(Y.shape)
```

```python
# Code Cell [6] Label 확인

Y
```

```python
# Code Cell [7] Label 변환

# Setosa      : +1
# Versicolor  : -1

Y = np.where(
    Y == 0,
    1,
    -1
)

Y
```

```python
# Code Cell [8] 표준화

scaler = StandardScaler()

X = scaler.fit_transform(X)

X[:10]
```

```python
# Code Cell [9] Weight / Bias

w = np.array([
    0.5,
    0.5,
    0.5,
    0.5
])

b = 0.1

print("Weight:", w)
print("Bias:", b)
```

```python
# Code Cell [10] Feed-Forward

z = X @ w + b

print("X shape:", X.shape)
print("w shape:", w.shape)
print("z shape:", z.shape)

z
```

```python
# Code Cell [11] Step Function

y_pred = np.where(
    z >= 0,
    1,
    -1
)

y_pred
```

```python
# Code Cell [12] 정확도

classified = np.sum(
    y_pred == Y
)

accuracy = np.mean(
    y_pred == Y
)

print(
    f"정분류 개수: "
    f"{classified} / {len(Y)}"
)

print(
    f"Accuracy: {accuracy:.3f}"
)
```

## Loss

실제값과 예측값의 차이를 확인합니다.

### Mean Squared Error (MSE)

$$
MSE = \frac{1}{N}\sum(y-\hat{y})^2
$$

### Mean Absolute Error (MAE)

$$
MAE = \frac{1}{N}\sum|y-\hat{y}|
$$

```python
# Code Cell [13] Loss

mse = np.mean(
    (Y - y_pred) ** 2
)

mae = np.mean(
    np.abs(Y - y_pred)
)

print(
    f"MSE = {mse:.3f}"
)

print(
    f"MAE = {mae:.3f}"
)
```
